## 4. The unit of analysis, as a real dataframe

**One row = one page.** Columns are observable signals about that page,
plus the target I would predict.*

```python
import os, sys, subprocess
import pandas as pd

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"
path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(path):
    path = f"{REPO_DIR}/{path}"
    if not os.path.exists(path):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

df = pd.read_csv(path)
df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)

cols = ["content_age_days", "days_since_last_update", "impressions_90d",
        "avg_position", "word_count", "is_declining"]
print(f"{len(df):,} rows. One row = one page.")
print(f"Share declining (target = 1): {df['is_declining'].mean():.1%}")
df[cols].head(10)
```

## 5. Why ML beats a fixed rule here
A fixed rule like "stale and visible" is easy to read, but it uses only two
signals and its thresholds are guesses. In the starter notebooks my hand rule found
about 12 of the top 50 declining pages, while a learned model found about 37.
A model can combine several signals and learn where the cutoffs actually are.
Output action: an editor reviews the top-ranked pages and decides which to update.
This is decision support, not proof of why a page is declining.
