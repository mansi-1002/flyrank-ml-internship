## 2. The question: decision, action, cost of a wrong call

***Question:** Which existing pages should a content team review first because
their search visibility is likely going down?

**Unit of analysis:** one page.
**Output:** a ranked list of pages to review, with the most likely declining pages first.

**Decision it improves:** which pages an editor spends limited time on this month.
**Who acts:** a content editor or SEO team member, who reviews and updates the top pages.

**Cost of a wrong call:** if a page on the list is actually fine, the editor wastes time.
If a page that really is declining is missed, it keeps losing visibility and nobody notices.
Wrong recommendations are cheap to check (an editor can open the page), but missed
pages are costly because the loss is silent.

**Why data or ML can help:** a team cannot read thousands of pages every month.
My own rule of thumb was only partly right, and a model can combine several signals
at once. The goal is not just to "train a model", but to give a short list that a
person can check and act on.

## 3. Quick look at the data (2-3 real numbers)
These numbers come from the starter CSV (anonymized, ~30,000 pages):

1. About [x]% of pages are labelled "down", so declining pages are common
   and worth finding.
2. The correlation between search_volume and impressions_90d is about [x],
   which is close to zero, so keyword popularity alone does not explain how visible a page is.
3. Median word count is similar for "down" and "up" pages ([x] vs [x]),
   so "write longer content" does not look like the main lever.

Together these suggest that a ranked list built from several signals could beat
simple intuition.

```python
import os, sys, subprocess
import pandas as pd

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

# Find the starter CSV, cloning the starter repo if needed (Colab)
path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(path):
    path = f"{REPO_DIR}/{path}"
    if not os.path.exists(path):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

df = pd.read_csv(path)

declining = df["trend_direction"].str.lower().eq("down").mean()
corr = df["search_volume"].corr(df["impressions_90d"])
med = df.groupby("trend_direction")["word_count"].median()

print(f"Pages: {len(df):,}")
print(f"1) Share of pages with trend_direction == 'down': {declining:.1%}")
print(f"2) Correlation search_volume vs impressions_90d: {corr:.3f}")
print(f"3) Median word count, down vs up: {med['down']:.0f} vs {med['up']:.0f}")
```
